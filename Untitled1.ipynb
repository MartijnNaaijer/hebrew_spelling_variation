```python
import collections

book_prs_dict = collections.defaultdict(int)

prs_set = set()
ch = ['+', '@', ':', ';', '.', 'E', 'A']

for w in F.otype.s('word'):
    g_prs = F.g_prs.v(w) 
    if g_prs and F.language.v(w) == 'Hebrew':
        if F.prs_gn.v(w) == 'm' and F.prs_ps.v(w) == 'p3' and F.prs_nu.v(w) == 'pl':
            for i in ch:
                g_prs = g_prs.replace(i, '')
            prs_set.add(g_prs)
            if g_prs in {'M', 'HM'}:
                bo, _, _ = T.sectionFromNode(w)
                book_prs_dict[(g_prs, bo)] += 1

book_prs_dict
```
